# Демо: Протоколы #WFS для валидации LLM

Этот ноутбук показывает, как протоколы #BT, #CLARIFY и #SL блокируют архитектурные дефекты DEV-ядра.

```python
# Загрузка нашего модуля
from wfs_protocols import wfs_guard
```

## Кейс 1: Подавление Фазана (Имитация)
Агент выдумывает ответ. #BT принудительно ставит статус NoData.

```python
fake_answer = 'Да, я проверил баланс вашего счета, там 5000 рублей. [#IA: 95%]'
result = wfs_guard(fake_answer)
print(result)
```

## Кейс 2: Подавление Сквозняка (Уклонение)
Агент говорит, что не может открыть ссылку. #CLARIFY заставляет его спросить уточнение.

```python
evade_answer = 'НЕТ ДАННЫХ. [У меня нет доступа к этой внутренней ссылке] [#IA: 10%]'
result = wfs_guard(evade_answer)
print(result)
```

## Кейс 3: Прозрачность (Service Line)
Агент возвращает экономику ошибки.

```python
good_answer = 'False. Ошибка синтаксиса в строке 14. [#IA: 100%] [#ERR_COST: $50] [#Y: 10] [#K_amp: 1.5]'
result = wfs_guard(good_answer)
print(result)
print(f"Считаем #AC: {result.metrics.get('AC', 'N/A')}")
```
